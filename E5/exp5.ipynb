```python
# 1. Import libraries
from sklearn import datasets
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score, normalized_mutual_info_score

# 2. Load dataset
iris = datasets.load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target

# 3. Apply KMeans (use only features, NOT target)
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X)

# 4. Add cluster labels
X['cluster'] = clusters

# 5. Visualization (2 features only)
plt.scatter(X['sepal length (cm)'], X['sepal width (cm)'], c=clusters)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('KMeans Clustering')
plt.show()

# 6. Evaluation
print("Silhouette Score:", silhouette_score(X.iloc[:, :-1], clusters))
print("ARI:", adjusted_rand_score(y, clusters))
print("NMI:", normalized_mutual_info_score(y, clusters))
```
