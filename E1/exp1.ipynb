```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
df = pd.read_csv('supermarket.csv')
```

```python
print(df.head())
print(df.info())
```

**Median**

```python
print("Median:\n", df.median(numeric_only=True))
```

**Categorical Modes**

```python
print("Product line mode:", df['Product line'].mode()[0])
print("City mode:", df['City'].mode()[0])
print("Payment mode:", df['Payment'].mode()[0])
```

**Scatter plots**

```python
plt.scatter(df['Quantity'], df['Total'])
plt.xlabel("Quantity")
plt.ylabel("Total")
plt.title("Quantity vs Total")
plt.show()
```

**Box Plot**

```python
df[['Tax 5%', 'gross income', 'Rating']].plot(kind='box')
plt.title("Boxplot")
plt.show()  
```

**Distribution plot**

```python
sns.histplot(df['Total'], kde = True)
plt.title("Distribution of Total")
plt.show()
```

```python
# 🔹 Trimmed Mean
trimmed_mean = stats.trim_mean(df['Total'], 0.1)
print("Trimmed Mean:", trimmed_mean)

# 🔹  Frequency Count
print(df['Product line'].value_counts())

# 🔹  Variance & Standard Error
print("\nVariance:\n", df.var(numeric_only=True))
print("\nSEM:\n", df.sem(numeric_only=True))

# 🔹  Correlation Matrix
print(df.corr(numeric_only=True))

# 🔹  Skewness & Kurtosis
print("\nSkewness:\n", df.skew(numeric_only=True))
print("\nKurtosis (Total):", df['Total'].kurtosis())

# 🔹 Sum of Squares
sos = np.sum(df['Total']**2)
print("\nSum of Squares:", sos)
```
